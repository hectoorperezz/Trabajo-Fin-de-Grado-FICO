```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Datos proporcionados
data = {
    'Promedio de las 36 acciones': [29.8, 2.35, 17, 117, 1.09, 5.39, 26.0, 25.0, 118.32, 1302019234],
    'Promedio del S&P500': [31.4, 2.336, 11, 104, 1.06, 4.76, 23.0, 26.3, 87.71, 9750875366]
}

variables = ['P/E', 'Rentabilidad Dividendo (%)', 'Price/Book', 'EV (€)', 'Beta', 'Volumen Promedio (€)',
             '% de propiedad de las "Big 3"', 'Volatilidad Anual (%)', 'BPA', 'Capitalización de mercado']

df = pd.DataFrame(data, index=variables)

fig, axes = plt.subplots(nrows=len(variables), ncols=1, figsize=(12, 20))

for i, ax in enumerate(axes):
    var = variables[i]
    ax.bar(['Promedio de las 36 acciones', 'Promedio del S&P500'], df.loc[var], color=['tab:blue', 'tab:orange'])
    ax.set_ylabel(var)
    ax.set_ylim(0, max(df.loc[var]) * 1.1)  # Aumentar un 10% el límite para mejor visualización
    for j, val in enumerate(df.loc[var]):
        ax.text(j, val, f'{val:.2f}', ha='center', va='bottom')

# Ajustar el layout
plt.tight_layout()
plt.show()
```
