```python
import random

# Generar 10 números aleatorios del 1 al 500
numeros_aleatorios = [random.randint(1, 500) for _ in range(10)]
print(numeros_aleatorios)
```

```python
import pandas as pd

# Reemplaza 'ruta_a_tu_archivo.csv' con la ruta real de tu archivo CSV
df = pd.read_csv('/content/csv_sp500list.csv', sep = ';')

print(df)

# Ahora df contiene tu DataFrame
```

```python
# Ajustar los índices para Python (base 0)
indices_ajustados = [n-1 for n in numeros_aleatorios if n-1 < len(df)]

# Filtrar las filas del DataFrame usando los índices ajustados
df_filtrado = df.iloc[indices_ajustados]

# Mostrar el DataFrame filtrado
print(df_filtrado)
```
