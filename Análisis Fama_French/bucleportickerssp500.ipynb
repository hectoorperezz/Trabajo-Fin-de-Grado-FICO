```python
import requests
import time

# Tu clave de API de Alpha Vantage
api_key = 'YOUR_API_KEY'

# Lista de símbolos de acciones (este es solo un ejemplo pequeño)
acciones = ['GOOGL', 'AAPL', 'MSFT']  # Supongamos que esta es una muestra de tu lista de 500 acciones

# URL base de Alpha Vantage
base_url = 'https://www.alphavantage.co/query?'

# Función para obtener datos de precios de cierre diario
def obtener_precios_cierre(api_key, simbolo):
    parametros = {
        'function': 'TIME_SERIES_DAILY',
        'symbol': simbolo,
        'apikey': api_key
    }
    response = requests.get(base_url, params=parametros)
    if response.status_code == 200:
        data = response.json()
        # Extraemos solo los datos de precios de cierre
        precios_cierre = {fecha: datos['4. close'] for fecha, datos in data['Time Series (Daily)'].items()}
        return precios_cierre
    else:
        print(f'Error al obtener datos para {simbolo}: {response.status_code}')
        return {}

# Ejemplo de cómo usar la función para múltiples acciones
for simbolo in acciones:
    precios = obtener_precios_cierre(api_key, simbolo)
    print(f'Precios de cierre para {simbolo}:', precios)
    time.sleep(12)  # Pausa para no exceder el límite de llamadas por minuto de la API
```
