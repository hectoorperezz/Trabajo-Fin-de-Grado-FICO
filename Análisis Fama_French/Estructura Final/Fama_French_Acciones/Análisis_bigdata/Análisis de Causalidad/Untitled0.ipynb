```python
import yfinance as yf
import pandas as pd

def get_major_holders_concentration(tickers):
    # Crear un DataFrame vacío con las columnas deseadas
    columns = ['insidersPercentHeld', 'institutionsPercentHeld', 'institutionsFloatPercentHeld', 'institutionsCount']
    results_df = pd.DataFrame(columns=columns)

    for ticker in tickers:
        stock = yf.Ticker(ticker)
        # Obtener los principales accionistas
        holders = stock.major_holders
        print(holders)
        if holders is not None and not holders.empty:
            # Extraer los datos en un diccionario
            # Asumiendo que el DataFrame tiene etiquetas de fila descriptivas y no índices numéricos
            data = {
                  'Insiders Percent Held': holders.at['Insiders Percent Held', 'Value'],
                  'Institutions Percent Held': holders.at['Institutions Percent Held', 'Value'],
                  'Institutions Float Percent Held': holders.at['Institutions Float Percent Held', 'Value'],
                  'Institutions Count': int(holders.at['Institutions Count', 'Value'])
              }
            row_df = pd.DataFrame([data], index=[ticker])
            # Append al DataFrame principal
            results_df = results_df.append(row_df)

    return results_df

# Lista de tickers para el ejemplo
tickers = ["AAPL", "MSFT", "AMZN", "GOOGL", "FB"]

# Ejecutar la función y mostrar los resultados
concentration_data = get_major_holders_concentration(tickers)
print(concentration_data)
```

### Resumen Extenso del Manual de Prevención de Blanqueo de Capitales y Financiación del Terrorismo del Grupo CASER

#### Introducción

El Grupo CASER ha adoptado medidas estrictas para prevenir la utilización del sistema financiero en el blanqueo de capitales y la financiación del terrorismo, basando sus políticas en el principio de **“Conoce a tu cliente” (KYC: Know Your Customer)**. Esta política de KYC se articula en tres pilares fundamentales:

1. **Política de Admisión de Clientes**
2. **Procedimientos de Diligencia Debida**:
   - **Identificación de clientes**
   - **Conocimiento de la relación de negocio**
   - **Seguimiento continuo de la relación de negocio**

La identificación y el conocimiento adecuado de los clientes es crucial para implementar eficazmente estas políticas. Es esencial obtener y verificar de manera diligente la información de identidad y actividad profesional de los clientes, en conformidad con los umbrales establecidos según el tipo de producto contratado y su riesgo asociado de blanqueo de capitales y financiación del terrorismo.

#### Definición de Cliente

Para efectos de la prevención del blanqueo de capitales, se considera **“cliente”** a cualquier persona física o jurídica en los siguientes momentos:
- **Contratación o alta del producto**: Incluye al tomador de la póliza, partícipe en los planes de pensiones y promotor en planes de pensiones de empleo.
- **Pago de prestaciones**: Incluye a los perceptores del importe en caso de rescate, anticipo, rentas, vencimiento, fallecimiento del asegurado, o cualquier otro concepto de pago.

#### Evaluación del Riesgo de Clientes

El Grupo CASER clasifica a sus clientes en función de los riesgos implícitos para prevenir eficazmente el blanqueo de capitales y la financiación del terrorismo. Esta evaluación considera factores como:

- **Segmentación de Clientes**: Basada en características específicas.
- **Características de los Productos**: Incluyendo seguros sobre la vida y planes de pensiones.
- **Operativas Relacionadas con Blanqueo y Financiación del Terrorismo**: Observando el uso de estos productos por parte de los clientes.
- **Otros Factores**: Incluyendo canales de distribución y otros elementos relevantes.

#### Política de Admisión de Clientes

La **Política de Admisión de Clientes** define varias categorías de clientes y establece procedimientos específicos para cada una. A continuación, se detallan las categorías y los criterios asociados:

1. **Clientes Excluidos de Aceptación**:
   - No se aceptan clientes que no sean previamente clientes de la entidad financiera correspondiente.
   - No se admiten clientes que no acepten la domiciliación bancaria o la transferencia para el pago de primas.
   - Personas que proporcionan datos falsos o incorrectos.
   - Personas residentes en paraísos fiscales o en países no cooperantes en la lucha contra el blanqueo de capitales.
   - Entidades financieras sin presencia física en su país de residencia.
   - Personas o entidades relacionadas con actividades terroristas o criminales según listas oficiales.

2. **Clientes con Alto Riesgo**:
   - **Personas con Responsabilidad Pública (PRPs)**: Incluye a aquellos que desempeñan o han desempeñado funciones públicas importantes.
   - Clientes con actividades, residencia o ingresos de alto riesgo.
   - Clientes que contratan productos de alto riesgo o superan ciertos umbrales monetarios.
   - Clientes relacionados con la producción o distribución de armas, comercio de arte y antigüedades.

3. **Clientes con Riesgo Normal**:
   - Menores de edad o mayores de 70 años.
   - Clientes cuyas operaciones no encajan claramente en otras categorías pero presentan sospechas.

4. **Clientes con Riesgo Bajo**:
   - Clientes que no cumplen con las características de alto o normal riesgo.

#### Procedimientos de Diligencia Debida

Los procedimientos de diligencia debida varían según el nivel de riesgo del cliente y el tipo de producto contratado:

1. **Diligencia Debida Normal**:
   - **Identificación Formal de Clientes**: Utilizando documentación acreditativa.
   - **Cumplimentación de Formularios Específicos**: Anexo 3 para personas físicas, Anexo 5 para personas jurídicas.
   - **Registro de Beneficiarios**: Identificación de beneficiarios previa al pago de prestaciones.

2. **Diligencia Debida Reforzada**:
   - Aplicable a clientes de alto riesgo o productos de alto riesgo.
   - Solicitud de documentación adicional sobre ingresos, actividad y origen de fondos.
   - Seguimiento reforzado de la relación de negocio y examen detallado de las operaciones.

3. **Diligencia Debida Simplificada**:
   - Aplicable a clientes de bajo riesgo o productos específicos.
   - Comprobación de identidad y propósito de la relación de negocio.
   - Menor requerimiento de información sobre la actividad profesional del cliente.

#### Procedimientos Específicos

1. **Identificación del Titular Real**:
   - Determinación y verificación del titular real en operaciones y relaciones de negocio.
   - Obtención de certificaciones y documentación adicional cuando sea necesario.

2. **Propósito e Índole de la Relación de Negocio**:
   - Recabar información sobre la naturaleza de la actividad del cliente y verificar su veracidad.
   - Utilización de anexos específicos para detallar la información empresarial o profesional del cliente.

3. **Archivo de Documentación**:
   - Almacenamiento de documentación en sistemas electrónicos y la Herramienta de Prevención del Grupo CASER.

#### Revisión y Actualización

La **UPBC** (Unidad de Prevención de Blanqueo de Capitales) es responsable de revisar y actualizar la **Política de Admisión de Clientes** anualmente o cuando las circunstancias lo requieran, sometiendo cualquier cambio a la aprobación del Consejo de Administración.

#### Distribución de la Política de Admisión de Clientes

Una vez aprobada, la política se distribuye a todos los empleados y canales de comercialización del Grupo CASER a través de la intranet, cursos de formación y otros medios específicos.

#### Procedimientos de Diligencia Debida Normal

Para clientes con riesgo promedio o inferior, se aplican los siguientes procedimientos:

1. **Identificación Formal del Cliente**:
   - Identificación previa a la relación de negocio mediante documentación acreditativa.
   - Cumplimentación de formularios específicos (Anexo 3 para personas físicas, Anexo 5 para personas jurídicas).
   - Solicitud de copia de la documentación acreditativa de la identificación del cliente.

2. **Identificación de Intervinientes en Productos Comercializados Telefónicamente o a través de Internet**:
   - Establecimiento de relaciones de negocio mediante medios telefónicos, electrónicos o telemáticos.
   - Acreditación de la identidad del cliente según normativa aplicable sobre firma electrónica o mediante procedimientos seguros autorizados.

3. **Documentación Identificativa del Cliente**:
   - Solicitud de fotocopia nítida o copia digital de la documentación escaneada para archivo y conservación.
   - Comprobación de la vigencia de la documentación aportada.

4. **Identificación del Titular Real**:
   - Identificación del titular real de la operación según la Ley 10/2010.
   - Recabar información para determinar si el cliente actúa por cuenta propia o de terceros.
   - Adopción de medidas para determinar la estructura de propiedad o de control de las personas jurídicas.

5. **Propósito e Índole de la Relación de Negocio**:
   - Recabar información sobre el propósito e índole de la relación de negocios con el cliente.
   - Verificación de la naturaleza de la actividad profesional o empresarial del cliente.

6. **Archivo de la Documentación**:
   - Almacenamiento de la documentación solicitada en la aplicación eDOCS y la Herramienta de Prevención.

#### Procedimientos de Diligencia Debida Reforzada

Para clientes de alto riesgo, se aplican medidas adicionales:

1. **Cumplimentación de Formularios Específicos**:
   - Anexo 6 para personas físicas y Anexo 7 para personas jurídicas cuando el importe de la prima sea superior a 150.000 euros.

2. **Recabar Documentación Adicional**:
   - Solicitud de documentación acreditativa de los ingresos declarados por el cliente.
   - Seguimiento reforzado de la relación de negocio y examen detallado de la lógica económica de las operaciones.

3. **Personas con Responsabilidad Pública (PRP)**:
   - Aplicación de medidas reforzadas para clientes con la condición de PRP, incluyendo la obtención de autorización previa y expresa del Director de Operaciones Vida y Pensiones.

#### Procedimientos de Diligencia Debida Simplificada

Para clientes y productos de bajo riesgo, se aplican medidas simplificadas:

1. **Clientes**:
   - Entidades de derecho público de los Estados miembros de la UE o países equivalentes.
   - Entidades financieras domiciliadas en la UE o países equivalentes bajo supervisión de cumplimiento de medidas de diligencia debida.
   - Sociedades con cotización en bolsa en la UE o países equivalentes.

2. **Productos u Operaciones**:
   - Pólizas de seguros de vida con primas anuales no superiores a 1.000 euros o primas únicas no superiores a

 2.500 euros.
   - Instrumentos de previsión social complementarios limitados a los supuestos contemplados en la normativa de planes y fondos de pensiones.

3. **Medidas Simplificadas**:
   - Comprobación de la identidad del cliente o titular real.
   - Cuestionario para la identificación de PRPs.
   - Inferencia del propósito y naturaleza del negocio por el tipo de producto contratado.

#### Conclusión

La implementación y el cumplimiento de estos procedimientos son esenciales para asegurar que el Grupo CASER actúe eficazmente en la prevención del blanqueo de capitales y la financiación del terrorismo, garantizando la integridad del sistema financiero y cumpliendo con las normativas vigentes. Este enfoque permite al Grupo CASER mitigar riesgos, mantener altos estándares de diligencia y fortalecer la confianza en sus operaciones y relaciones de negocio.
