```python
import yfinance as yf
import pandas as pd
import numpy as np

# Assuming df_sectores is already defined and contains the necessary data
lista_tickers_tecnologia = df_sectores[df_sectores['Sector'] == 'Consumer Cyclical']
tech_tickers = lista_tickers_tecnologia['Ticker'].tolist()

# Imprimir la lista de tickers
print("Tickers del sector Technology:", tech_tickers)

def rentabilidad_sector(lista_tickers_tecnologia):
    # Definir el rango de fechas
    fecha_inicio = '2020-01-01'
    fecha_fin = '2023-12-31'

    # DataFrame para almacenar los precios de cierre semanales y market cap
    df_sector = pd.DataFrame()

    try:
        for ticker in lista_tickers_tecnologia:
            stock = yf.Ticker(ticker)
            # Obtener datos históricos semanales de precios
            data = stock.history(start=fecha_inicio, end=fecha_fin, interval='1wk')
            data = pd.DataFrame(data)
            data = data['Close']
            data = pd.DataFrame(data, columns=['Close'])
            data = data.reset_index()
            data['Date'] = pd.to_datetime(data['Date'])  # Convert to datetime
            data['Date'] = data['Date'].dt.strftime('%d-%m-%Y')
            data.set_index('Date', inplace=True)

            # Obtener el número de acciones usando el rango completo y luego reducir a datos semanales
            shares_data = stock.get_shares_full(start=fecha_inicio, end=fecha_fin)
            shares_data = pd.DataFrame(shares_data, columns=['shares'])
            shares_data = shares_data.reset_index()
            shares_data['index'] = pd.to_datetime(shares_data['index'])  # Convert to datetime
            shares_data['index'] = shares_data['index'].dt.strftime('%d-%m-%Y')
            shares_data.set_index('index', inplace=True)
            shares_data = shares_data[~shares_data.index.duplicated(keep='last')]

            # Alinear shares_data con data usando el índice de data
            shares_data = shares_data.reindex(data.index)
            shares_data.fillna(method='ffill', inplace=True)  # Rellenar hacia adelante

            market_cap = pd.merge(data, shares_data, left_index=True, right_index=True, how='left')

            # Calcular la capitalización de mercado
            market_cap['Market Cap'] = market_cap['Close'] * market_cap['shares']

            # Calcular rendimientos logarítmicos
            market_cap['Log Close'] = np.log(market_cap['Close'])
            market_cap['Log Returns'] = market_cap['Log Close'].diff()
            market_cap[f'{ticker}_Market Cap'] = market_cap['Market Cap']
            market_cap[f'{ticker}_Log Returns'] = market_cap['Log Returns']
            df_final = market_cap[[f'{ticker}_Market Cap', f'{ticker}_Log Returns']].copy()

            market_cap_subset = df_final[[f'{ticker}_Market Cap', f'{ticker}_Log Returns']].copy()

            # Concatenar los resultados de este ticker al DataFrame final
            df_sector = pd.concat([df_sector, market_cap_subset], axis=1)

        # Imprimir el DataFrame resultante
        print(df_sector)

        # Calcular la rentabilidad ponderada
        total_market_cap = df_sector.filter(like='Market Cap').sum(axis=1)
        for ticker in lista_tickers_tecnologia:
            df_sector[f'{ticker}_Weighted Returns'] = df_sector[f'{ticker}_Log Returns'] * (df_sector[f'{ticker}_Market Cap'] / total_market_cap)

        df_sector['Weighted Returns'] = df_sector.filter(like='Weighted Returns').sum(axis=1)
        df_final = df_sector['Weighted Returns']

        # Imprimir el DataFrame final
        print(df_sector[['Weighted Returns']])

    except Exception as e:
        print(f"Error al procesar {ticker}: {e}")
        df_final = pd.DataFrame()  # Return an empty DataFrame in case of an error

    return df_final

rent_sector = rentabilidad_sector(tech_tickers)
rent_sector.to_csv('rent_sector.csv', index=False)
```
