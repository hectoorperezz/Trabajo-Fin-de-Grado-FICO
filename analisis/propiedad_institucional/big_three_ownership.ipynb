```python
import yfinance as yf
import pandas as pd

def obtener_porcentaje_institucional(ticker):
    try:
        stock = yf.Ticker(ticker)
        holders = stock.get_institutional_holders()

        if holders is not None and not holders.empty:
            filtro = holders['Holder'].isin(['Vanguard Group Inc', 'Blackrock Inc.', 'State Street Corporation'])
            suma_porcentaje = holders.loc[filtro, 'pctHeld'].sum()  # Asegúrate que 'pctHeld' es el nombre correcto
            return suma_porcentaje
        else:
            print(f"No se encontraron datos institucionales suficientes para {ticker}.")
            return None
    except Exception as e:
        print(f"Error al procesar los datos para {ticker}: {e}")
        return None

# Lista de tickers de interés
tickers = ["NCLH",
"TRGP",
"WYNN",
"RCL",
"GEN",
"META",
"CLX",
"SWKS",
"QRVO",
"NKE",
"IEX",
"FAST",
"ADM",
"BAC",
"MSI",
"ANSS",
"CMCSA",
"ALLE",
"CSCO",
"COST",
"AOS",
"JPM",
"ABBV",
"AMP",
"NWSA",
"AMT",
"ACN",
"UNH",
"LUV",
"CCI",
"JKHY",
"MCK",
"DLR",
"COR",
"MOH",
"CNC",]


# Crear una lista para guardar temporalmente los DataFrames
dataframes = []

# Iterar sobre cada ticker
for ticker in tickers:
    porcentaje = obtener_porcentaje_institucional(ticker)
    if porcentaje is not None:  # Solo añadir al DataFrame si hay un porcentaje válido
        df_temp = pd.DataFrame({'Ticker': [ticker], 'Percent Held by Institutions': [porcentaje]})
        dataframes.append(df_temp)

# Concatenar todos los DataFrames temporales en uno solo si hay datos
if dataframes:
    resultados = pd.concat(dataframes, ignore_index=True)
    print(resultados)

    # Exportar los resultados a un archivo CSV
    resultados.to_csv('institutional_holders_percentage.csv', index=False)
    print("Los datos han sido exportados a 'institutional_holders_percentage.csv'.")
else:
    print("No se obtuvieron datos válidos para ninguno de los tickers.")
```
