```python
import yfinance as yf
import pandas as pd
import numpy as np

tickers_a_buscar = ["PH", "BXP", "FSLR", "CBRE", "DAL", "CI", "MMM", "UDR", "ALLE", "EQR",
                    "GD", "BAX", "MCHP", "HUBB", "UAL", "CAH", "STT", "GEN", "AAL", "BK",
                    "ILMN", "PFG", "RJF", "ULTA", "AXP", "ETN", "WAB", "CPT"]

# Descargar datos diarios
data = yf.download(tickers_a_buscar, period="1d", group_by='ticker')

info_frames = []

for ticker in tickers_a_buscar:
    stock = yf.Ticker(ticker)
    info = stock.info
    summary = pd.DataFrame({
        'Ticker': [ticker],
        'Name': [info.get('longName', np.nan)],
        'Sector': [info.get('sector', np.nan)],
        'Industry': [info.get('industry', np.nan)],
        'Trailing P/E': [info.get('trailingPE', np.nan)],
        'Forward P/E': [info.get('forwardPE', np.nan)],
        'Price/Sales': [info.get('priceToSalesTrailing12Months', np.nan)],
        'Price/Book': [info.get('priceToBook', np.nan)],
        'Dividend Yield': [info.get('dividendYield', np.nan) * 100 if info.get('dividendYield') else np.nan],
        'Profit Margin': [info.get('profitMargins', np.nan)],
        'Operating Margin': [info.get('operatingMargins', np.nan)],
        'Return on Assets': [info.get('returnOnAssets', np.nan)],
        'Return on Equity': [info.get('returnOnEquity', np.nan)],
        'Revenue Per Share': [info.get('revenuePerShare', np.nan)],
        'Market Cap': [info.get('marketCap', np.nan)],
        'Enterprise Value': [info.get('enterpriseValue', np.nan)],
        'Beta': [info.get('beta', np.nan)],
        'Average Volume': [info.get('averageVolume', np.nan)],
        'Shares Outstanding': [info.get('sharesOutstanding', np.nan)],
        'Float': [info.get('floatShares', np.nan)],
        'Percentage Held by Insiders': [info.get('heldPercentInsiders', np.nan)],
        'Percentage Held by Institutions': [info.get('heldPercentInstitutions', np.nan)],
        'Shares Short': [info.get('sharesShort', np.nan)],
        'Short Ratio': [info.get('shortRatio', np.nan)],
        'Short % of Float': [info.get('shortPercentOfFloat', np.nan)],
        'Short % of Shares Outstanding': [info.get('shortPercentOfSharesOutstanding', np.nan)],
        'Earnings Quarterly Growth': [info.get('earningsQuarterlyGrowth', np.nan)],
        'Date': [info.get('lastFiscalYearEnd', np.nan)],
        'Most Recent Quarter': [info.get('mostRecentQuarter', np.nan)],
        'Next Earnings Date': [info.get('earningsAnnouncement', np.nan)]
    })
    info_frames.append(summary)

info_df = pd.concat(info_frames, ignore_index=True)
info_df.to_csv('stock_information_detailed.csv', index=False)
print("Detailed stock information has been exported successfully to 'stock_information_detailed.csv'.")
```
